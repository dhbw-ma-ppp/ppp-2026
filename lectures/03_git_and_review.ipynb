# git, pull requests and code review

## Why version control?

- **History:** every saved state of your project, with who changed what, when and why
- **Undo:** go back to any earlier state
- **Collaboration:** many people work on the same code at the same time, without overwriting
  each other's work
- **Review:** changes are discussed *before* they become part of the project

**git** is the version control system practically everyone uses (Linus Torvalds, 2005).
**GitHub** is a website that hosts git repositories and adds pull requests, reviews and CI.
You will very likely use both in your company from day one.

## Where your changes live

```
 working directory ──git add──▶ staging area ──git commit──▶ local history ──git push──▶ GitHub
  (your files)                  (next commit)                 (on your laptop)            (remote)
                                                               ◀──────────git pull─────────
```

- **Working directory:** the files you edit
- **Staging area:** what goes into the next commit. You choose, file by file, with `git add`
- **Commit:** a snapshot of the staged files, with a message, an author and a time
- **Push / pull:** copy commits to / from GitHub

`git status` tells you, at any time, what is where. Use it constantly.

## Commits and branches

```
main:                A ── B ── C
                               └── D ── E      ex01-alice
```

- Each commit points to the commit before it: the history is a chain.
- A **branch** is just a **name pointing to a commit**. (Remember: names are labels!)
  When you commit on a branch, the name moves forward to the new commit.
- Each of you works on your own branch, so nobody gets in anyone's way.
- After the work has been reviewed, the branch can be **merged** into `main`.

## Pull requests

A **pull request (PR)** asks: "please merge my branch into `main`". On GitHub, it's the
place where

- **CI** runs automatically on every push: in our course, it checks that your files are in
  your folder, runs **ruff** (likely bugs, formatting) and runs **your tests**.
  Red means something failed: click *Details* and read the error.
- **Reviewers** read your changes, comment on lines, and ask questions.
- The author **answers and updates** the PR by pushing more commits to the same branch.
- Finally, someone merges it.

## Things that save you

- `git status`: before and after every other git command
- `git diff`: what did I change? `git diff --staged`: what am I about to commit?
- `git log --oneline`: the history of your branch
- Add your **folder**, not everything: `git add students/<login>/ex01` (or maybe even better, add individual files)
- A commit message says **why**, not just what: "ex01: handle empty input", not "changes"
- Stuck, or git says something scary? Don't force anything. `git status`, and ask.

## Code review

Reading other people's code is a large part of a programmer's job, and a skill you can
practise. A review:

- **catches bugs** before they reach `main`, especially in cases nobody tested
- **spreads knowledge**: you see other ways to solve the same problem
- **makes code understandable**: if the reviewer can't follow it, neither can your future self

What to look for:

- **Correctness:** does it do what it should, also for unusual input?
- **Tests:** is there a case that isn't tested?
- **Simplicity:** could it be simpler, or clearer?
- **Names:** do names say what things are?

Leave formatting to ruff; CI checks it already.

## Our review rules

A review counts only if it contains:

1. At least one **concrete suggested test case** (input, expected result, why it's interesting)
2. At least one **genuine question** about a design choice

Authors answer every review of their PR; that answer is part of their own PR.

Be specific, and kind: comment on the code, not the person.
Template and details: `REVIEWING.md`.

## Pair programming

For Part 1 in class you may work in pairs:

- **Driver:** types, and thinks about the next line
- **Navigator:** reads along, thinks ahead, spots mistakes, keeps the goal in mind
- **Switch roles** every 15 minutes

Both of you submit your own PR, and both of you must be able to explain it.
