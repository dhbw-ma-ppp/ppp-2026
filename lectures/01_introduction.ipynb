## What this course is about

In your companies you will soon work on real code, written by colleagues and, increasingly,
by AI assistants. An AI can solve every take-home exercise in this course in seconds.

So what is worth learning? The things that stay scarce:

- **Knowing what code does** when it runs, without running it
- **Stating a problem precisely**, before and while solving it
- **Testing**: convincing yourself (and others) that code is correct
- **Reading and reviewing** other people's code, critically
- **Breaking a problem down** into pieces you can solve
- **Explaining your own code**, and changing it on request

Everything in this course (exercises, reviews, the exam) trains these.

## Live review on Fridays

- Two PRs submitted the day before are picked **at random**.
- The **author** shows the code on the projector.
- The **assigned reviewer** asks their question or suggests their test case, live.
- The author answers, or (occasionally) makes the change on the spot.

Not graded, and not meant to embarrass anyone. It's how programmers work together.
But anyone can be picked, as author **or** as reviewer. So understand your own code,
and look at the PR you review before Friday.

## Code review

Each week you review one PR of a fellow student (assigned automatically, rotating).
A review counts only if it contains:

1. At least one **concrete suggested test case** (input, expected result, why it's interesting)
2. At least one **genuine question** about a design choice
3. An **answer from the author** in the PR thread

Reviews are due **Tuesday 17:00**; authors answer by the **Wednesday session**.

Details and a template: `REVIEWING.md` in the course repository.

## AI policy

Every exercise states its **AI level**:

- **Level 0:** no AI
- **Level 1:** AI may explain things, but writes no code
- **Level 2:** AI allowed; say in your PR how you used it

Weeks 1–4 are **Level 0**, and AI completion is switched off in VS Code for this repository.
First you build a mental model of how code runs; after that, AI becomes a useful tool
instead of a crutch. From week 5 on, we use AI in structured ways, e.g. you write the tests,
the AI writes the code, and you explain it.

## The exam

Paper, 90 minutes, Fri Dec 18. It asks for what you practise every week:

- **Trace** code and predict its output
- **Find the bug** in a given function
- **Write test cases** for a specification (as a table: input, expected output, what it covers)
- **Review** a short piece of code

You will not have to write programs by hand.

## Installing

**git**
- Windows: https://git-scm.com/downloads (defaults are fine)
- macOS: run `git --version` in a terminal; if missing, macOS offers to install it
- Linux: via your package manager, e.g. `sudo apt install git`

**VS Code**: https://code.visualstudio.com/ (on Windows the "User Installer" needs no admin rights)

**uv**: https://docs.astral.sh/uv/getting-started/installation/
- Windows (PowerShell): `powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"`
- macOS / Linux: `curl -LsSf https://astral.sh/uv/install.sh | sh`

After installing, **close and reopen** your terminal (and VS Code), so it finds the new programs.

## Configuring git and getting the repository

In a terminal (in VS Code: *Terminal → New Terminal*):

