# Part 1: The shape of a block

## One shape for everything new today

So far, every statement was a single line: an assignment, a `print`, a function call.
Today's new statements, `if`, `for`, `while` and `def`, are different: they **control other
statements**. Whether those run, how often, or when. And they all have the same shape:

```
header:
    block
    block
another header:
    block
```

- A **header** starts with a keyword (`if`, `elif`, `else`, `for`, `while`, `def`) and ends
  with a **colon**.
- Its **block** is all the lines below it that are indented one level deeper (4 spaces).
  The block ends at the first line that is indented less.
- Headers that belong together, like `if` / `elif` / `else`, are at the **same** indentation.
- A block can contain further headers with their own blocks.

(Python's documentation calls a header with its block a **clause**, and the block a **suite**.)

## Indentation is the structure

Many languages mark where a block starts and ends with symbols or keywords, and the
indentation is only there for the reader. With curly braces:

```
if (x > 0) {
    print("positive");
}
```

With keywords:

```
if x > 0 then
    print("positive")
end
```

In Python the indentation **is** the structure. What you see is what runs.

**Predict:** what does this print?

# Part 3: Repeating

## `for`: once for every element

````python
for char in "(()":
    print(char)
````

````python
for number in [1, 3, 5, 7]:
    print(number, 2 * number)
````

`for <name> in <iterable>:` runs its block once for every element of the iterable, meaning
anything Python can go through one element at a time (a string, a list, a `range`, ...).
`<name>` labels the current element.
That's Santa's elevator from Exercise 1, step by step:

````python
floor = 0
for char in "(()))":
    if char == "(":
        floor = floor + 1
    elif char == ")":
        floor = floor - 1
print(floor)
````

**Predict:** what changes if the `print` is indented by 4 more spaces? Try it.

This is a pattern you'll use all the time, **accumulating**: a name set *before* the loop,
updated *in* the loop, used *after* it.

### Aside: iterable vs. sequence

`for` only needs an **iterable**: something that hands out its elements one at a time.
A **sequence** is an iterable that can do more: it has a length, and you can index into it.
Strings, lists and `range` are sequences. `enumerate` is only an iterable: it produces the
pairs one by one, as the loop asks for them. Later you'll loop over more iterables that
aren't sequences: dictionaries, sets, the lines of a file

**Predict:** which of these lines work?

````python
numbers = range(5)
print(len(numbers), numbers[2])
pairs = enumerate("(()")
print(pairs[0])
````

If you need a sequence, `list(...)` makes one, as we did with `range` above:
`list(enumerate("(()"))[0]` works.

## Exercise 1 bonus: when does Santa first reach the basement?

Before writing code, write down examples: input, expected result, and which case each covers.

| Input | Expected | Case |
|---|---|---|
| `")"` | `1` | the very first step |
| `"()())"` | `5` | after going up and down |
| `"((("` | ? | never reaches the basement |

The last row is a question the description doesn't answer. We have to decide before we can
write the code. We decide: `None`, Python's value for "nothing".

````python
def first_basement(instructions):
    floor = 0
    for position, char in enumerate(instructions, start=1):
        if char == "(":
            floor = floor + 1
        else:
            floor = floor - 1
        if floor == -1:
            return position
    return None


print(first_basement(")"))
print(first_basement("()())"))
print(first_basement("((("))
````

**Predict:** for `"()())((("`, how often does the loop run?

`return` ends the function immediately, in the middle of the loop. A loop can't do that on its
own: this is something `count` couldn't do for us.

## `while`: repeat as long as a condition is true

`while <condition>:` checks the condition, runs the block if it is true, and then checks again.
`for` runs once per element; `while` is for when you don't know in advance how many rounds it
takes.

**Predict:** trace it on paper, with one row per round (`n`, `steps`). What is printed?

````python
n = 6
steps = 0
while n != 1:
    if n % 2 == 0:
        n = n // 2
    else:
        n = 3 * n + 1
    steps = steps + 1
print(steps)
````

(Nobody knows whether this loop ends for *every* starting number. It has been checked for all
numbers up to about 10²⁰.)

**Predict:** what does this loop do?

````python
n = 5
while n != 0:
    n = n - 2
````

## `break` and `continue`

**Predict:** which lines are printed?

````python
for number in range(6):
    if number % 2 == 0:
        continue
    print(number)
````

`continue` skips the rest of the block **for this round** and goes on with the next one.

**Predict:** what does this print?

````python
n = 0
while n < 5:
    if n == 2:
        continue
    print(n)
    n = n + 1
````

`0`, `1`, and then nothing, forever: when `n` is 2, `continue` skips `n = n + 1`, so `n` stays 2.
With `while`, make sure every path through the block brings the loop closer to its end.

# Part 4: Functions, properly

## `def`: the block runs when the function is called

`def <name>(<parameters>):` gives its block a name. **Predict:** what is printed, in which
order?

````python
def say_hello():
    print("hello")


print("defined")
````

Defining a function runs nothing. Its block runs each time the function is **called**.

## Parameters and arguments

The names in the header are the **parameters**: `def describe(name, age):`. The values in
the call are the **arguments**: `describe("Ada", 36)`.

A call first evaluates the arguments, then puts each parameter label on its argument, in
order, as if the block began with `name = "Ada"` and `age = 36`. Then the block runs.

**Predict:** what does each call print?

````python
def describe(name, age):
    print(name, "is", age)


describe("Ada", 36)
describe("A" + "da", 30 + 6)
describe(36, "Ada")
describe("Ada")
````

- Arguments are matched to parameters **by position**: the first argument goes to the first
  parameter. Python doesn't check that this makes sense; `describe(36, "Ada")` runs.
- Every parameter needs an argument, or the call fails before the block runs.
- Each call puts the labels anew: the block can run with different values each time.

## `return` or `print`?

**Predict:** what is printed, and in which order?

````python
def double_print(x):
    print(2 * x)


def double_return(x):
    return 2 * x


a = double_print(5)
b = double_return(5)
print(a, b)
````

- `print` shows a value **to a person**. The program can't use it afterwards.
- `return` hands a value **back to the code that called** the function.
- A function without `return` returns `None`. That's what `assert None == 212` in the test
  output meant last week.

Tests only see what you `return`.

## `None`: no value

````python
nothing = None
print(nothing)
print(type(nothing))
print(nothing is None)
````

Check for it with `is None` (or `is not None`), not with `==`.

`nothing == None` would also give `True` here. So why `is`? The two ask different questions:

- `a == b`: do `a` and `b` have **equal values**? Each type decides what "equal" means.
- `a is b`: are `a` and `b` labels on **the same object**?

There is only one `None` object, so `is None` asks exactly what we mean, and no type can
change the answer. With `==`, the type gets a say: in pandas (later in the course), `== None`
doesn't give you a single `True` or `False` at all. ruff in CI also flags `== None` (E711).

For anything other than `None`, compare with `==`. `is` on numbers or strings sometimes
gives `True` and sometimes doesn't.

## `return` ends the function

**Predict:** what is printed?

## Default values and keyword arguments

You've used this already: `enumerate("(()", start=1)`. A parameter can have a
**default value**, used when the caller leaves it out. **Predict:** what is printed?

````python
def greet(name, greeting="Hello", punctuation="!"):
    return f"{greeting}, {name}{punctuation}"


print(greet("Ada"))
print(greet("Ada", "Hi"))
print(greet("Ada", punctuation="?"))
print(greet(greeting="Moin", name="Ada"))
````

- Parameters with a default can be left out; the ones without come first and are required.
- A **keyword argument** (`punctuation="?"`) names the parameter it is for, instead of
  going by position. So you can skip some, or change the order.
- Use keywords when a bare value wouldn't tell the reader what it means:
  `enumerate(text, start=1)` reads better than `enumerate(text, 1)`.

## Two results: return a pair

**Predict:** what is printed?

## Names inside a function stay inside

**Predict:** what happens?

````python
def area(width, height):
    result = width * height
    return result


print(area(3, 4))
print(result)
````

Parameters (`width`, `height`) and names created inside a function only exist while it runs.
A function gets its input through its parameters, and gives back its result with `return`.

## Reading a name from outside, and assigning to it

**Predict:** what happens at each of the two calls?

````python
level = 3


def show_level():
    print(level)


def go_up():
    level = level + 1
    print(level)


show_level()
go_up()
````

A function can **read** a name from outside. But if it **assigns** to a name anywhere in
its block, that name is local in the *whole* block, so `level + 1` looks for a local
`level` that has no value yet.

Don't reach outside. Pass the value in, and `return` the new one:

````python
def go_up(level):
    return level + 1


level = go_up(level)
print(level)
````

## Functions use functions

Exercise 1, Part 2: `floor_after` was meant to use `floor`.

````python
def floor(instructions):
    """Return the floor Santa ends on: "(" is one up, ")" one down."""
    return instructions.count("(") - instructions.count(")")


def floor_after(instructions, steps):
    """Return the floor after the first `steps` instructions."""
    return floor(instructions[:steps])


print(floor_after("((())", 3))
````

Small functions that each do **one** thing can each be tested on their own, and each name tells
the reader what that piece does. Exercise 2 is built like this: `step` executes one
instruction, and `run` uses `step`.

## Who runs when

**Predict:** which lines are printed, in which order?

````python
def inner(x):
    print("inner starts, x =", x)
    return x * 2


def outer(x):
    print("outer starts, x =", x)
    y = inner(x + 1) + 1
    print("outer ends, x =", x)
    return y


print(outer(5))
````

A call **pauses** the caller. The called function runs until its `return`, and the caller
continues where it stopped, with the returned value in place of the call.
Each call has its own names: `inner`'s `x` is 6 while `outer`'s `x` stays 5.

*Watch it on pythontutor.com: each call gets its own box of names (a **frame**).*

## Docstrings: what a function promises

The text in triple quotes right after the header of `floor` and `floor_after` above is a
**docstring**. Python keeps it, and `help` shows it (so does VS Code, when you hover over
a call):

````python
help(floor_after)
````

A docstring says **what** the function promises (what goes in, what comes out, special
cases such as `None`), not how it works. The Exercise 2 stubs have one for every
function, and in Part 2 you'll write into one yourself.

Docstring and tests together are the specification: one says it in words, the other in
examples.

# Part 5: Tests as specification

## A test is an example of what a function should do

The table for `first_basement` above *was* the specification, and writing it found a question
(`"((("`) before we wrote any code. Each row becomes a test:

````python
def test_first_basement_first_step():
    assert first_basement(")") == 1


def test_first_basement_never():
    assert first_basement("(((") is None


test_first_basement_first_step()
test_first_basement_never()
print("both passed")
````

In a file named `test_....py`, `pytest` finds every function whose name starts with `test_`,
calls it, and reports each `assert` that fails. Here we call them by hand.

Good test cases each cover a **different case**: the first and the last position, never, empty
input. Ten tests of the same kind find no more than one.
