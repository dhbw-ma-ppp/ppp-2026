# Part 1: How a program runs

## What's inside a computer

## What's a programming language?

> A precisely defined way of writing text, so that a computer can execute it.

Python is **interpreted**: a program (the Python interpreter) reads your code and executes it,
**statement by statement, top to bottom**.

Two ways to run Python:
- **Interactively** (the REPL): `uv run python`, type a line, see the result
- **As a script**: save code in `hello.py`, run `uv run python hello.py`

## Top to bottom

**Predict:** what does this cell print?

````python
print("one")
print("two")
x = 5
print("three", x)
````

**Predict:** and this one?

````python
print("one")
print("two", y)
y = 5
print("three", y)
````

- Statements run **in order**. A name must exist *before* it is used.
- An error **stops the program** right there. Everything before it has already happened. Nothing after the failing statement gets executed
- The error message tells you *what* went wrong and *where*: read it from the bottom up.

## Syntax and semantics

- **Syntax**: the rules for which text is valid code at all
- **Semantics**: what valid code *means*, i.e. what it does

"Blue lemons flow particularly high." is valid English syntax, but meaningless.

**Predict:** what does this cell print?

````python
print("one")
print("two")
print("three" "four")
print("five" 5)
````

Nothing at all is printed: Python first checks the **whole** cell (or file) for syntax errors,
and only then starts executing. A `SyntaxError` means nothing ran.

And `"three" "four"`? That's valid, and means `"threefour"`. Valid syntax, surprising semantics.

## Syntactically fine, semantically wrong

The worst kind of bug: Python runs it without complaint.

````python
# the area of a rectangle
width = 3
height = 4
area = width + height
print("area:", area)
````

Python can't know what you *meant*. Only you can, and your tests.

# Part 2: Values and types

## Every value has a type

A **value** is a piece of data: `42`, `3.14`, `"hello"`, `True`.
Its **type** says what kind of data it is, and what you can do with it.

````python
print(type(42))
print(type(3.14))
print(type("hello"))
print(type(True))
````

| Type | Examples | |
|---|---|---|
| `int` | `42`, `-7`, `0` | whole numbers, **any size** |
| `float` | `3.14`, `-0.5`, `1e6` | numbers with a decimal point (approximate!) |
| `str` | `"hello"`, `'x'`, `""` | text |
| `bool` | `True`, `False` | truth values |

**Predict:** does this work, and if so, what does it print?

````python
print(2**100)
````

**Predict:** `True` or `False`?

````python
print(0.1 + 0.2 == 0.3)
````

````python
print(0.1 + 0.2)
````

Floats are stored in binary with limited precision, so most decimal fractions can only be
**approximated**. Never compare floats with `==`; ints are exact.

## Converting between types

````python
print(int("42") + 1)
print(str(42) + "1")
print(float("3.5"))
print(int(3.9))
````

**Predict:** what happens here?

````python
print("3" + 5)
````

Python does not guess: adding text and a number is a `TypeError`. Convert explicitly.

# Part 3: Expressions

## An expression is evaluated to a value

`3 + 4 * 2` is an **expression**. Python evaluates it, step by step, to a single value.

**Predict:** what's the value of each line?

````python
print(3 + 4 * 2)
print((3 + 4) * 2)
print(2**3**2)
print(-2**2)
````

Precedence, highest first: `**`, then unary `-`, then `*` `/` `//` `%`, then `+` `-`.
`**` groups from the right: `2**3**2` is `2**(3**2)`.
When in doubt, use parentheses. They cost nothing and help the reader.

## Division

- `/` : division, **always** gives a float
- `//` : integer division, rounds **down**
- `%` : remainder (modulo)

**Predict:** fill in the values

````python
print(7 / 2)
print(6 / 2)
print(7 // 2)
print(7 % 2)
print(-7 // 2)
print(-7 % 2)
````

Rounds **down** means towards minus infinity: `-7 // 2` is `-4`, not `-3`.
And always: `(a // b) * b + a % b == a`.

`divmod(a, b)` gives both at once:

````python
print(divmod(17, 5))
````

## Operators on other types

Some operators work on strings too, with their own meaning.

**Predict:** which lines work, and what do they print?

````python
print("ab" + "cd")
print("ab" * 3)
print("ab" * "cd")
````

## Comparisons give a `bool`

````python
print(3 < 5)
print(3 == 3.0)
print("apple" < "banana")
print(1 < 2 < 3)
````

- `=` **assigns**, `==` **compares**
- `and`, `or`, `not` combine booleans: `x > 0 and x < 10`

# Part 4: Names and objects

## Assignment binds a name to a value

```python
x = 3 + 4
```

1. **First** the right-hand side is evaluated: `7`
2. **Then** the name `x` is bound to that value

A name is a **label** attached to an object, not a box that contains it.

**Predict:** what's printed?

````python
x = 5
x = x + 1
print(x)
````

**Predict:** what's printed?

````python
a = 1
b = a
a = 2
print(a, b)
````

`b = a` attaches the label `b` to the object that `a` is attached to (the `1`).
`a = 2` then moves the label `a` to a new object. `b` still points to `1`.

*Try it on pythontutor.com.*

**Predict:** what's printed? (Swapping two values)

````python
a = 1
b = 2
a = b
b = a
print(a, b)
````

````python
# Python's way to swap: the right-hand side is evaluated completely first
a = 1
b = 2
a, b = b, a
print(a, b)
````

## Summary of part 1

- Code runs **top to bottom**, one statement at a time; an error stops it.
- Every value has a **type**; the type decides what operations mean.
- An **expression** is evaluated to a single value.
- **Assignment** first evaluates the right side, then binds the name on the left to it.
- A **name** is a label on an object.

# Part 5: Strings

## Strings are sequences of characters

````python
s = "Python"
print(len(s))
print(s[0])
print(s[1])
print(s[-1])
````

- Positions (**indices**) start at **0**; the last index is `len(s) - 1`
- Negative indices count from the end: `-1` is the last character

**Predict:** what happens?

````python
s = "Python"
print(s[6])
````

## Slicing: `s[start:stop]`

From `start` (included) up to `stop` (**excluded**).

**Predict:** fill in each line

````python
s = "Python"
print(s[0:2])
print(s[2:5])
print(s[:2])
print(s[2:])
print(s[-3:])
print(s[::2])
print(s[::-1])
````

- Leave out `start` or `stop` to mean "from the beginning" / "to the end"
- `s[:k] + s[k:] == s`, always, which is why `stop` is excluded
- A third number is the step: `s[::-1]` walks backwards
- Slicing never raises an error, even out of range: `s[2:100]` is fine

## Strings can't be changed

**Predict:** what's printed?

````python
s = "DataScience"
s.replace("Science", "Analytics")
print(s)
````

String methods never change the string; they **return a new one**.
To keep the result, bind a name to it:

````python
s = "DataScience"
s = s.replace("Science", "Analytics")
print(s)
````

**Predict:** what happens?

````python
s = "Python"
s[0] = "J"
````

## Useful string methods

````python
s = "Hello, World"
print(s.upper())
print(s.lower())
print(s.count("l"))
print(s.find("World"))
print(s.startswith("Hell"))
print("World" in s)
print(s.split(", "))
````

All methods: https://docs.python.org/3/library/stdtypes.html#string-methods
(or type `s.` in VS Code and look at the suggestions)

## f-strings: putting values into text

````python
name = "Ada"
year = 1815
print(f"{name} was born in {year}, {2026 - year} years ago.")
print(f"{year=}")
````

Anything inside `{}` is an expression, evaluated and inserted.
`{year=}` shows name and value, handy while debugging.

# Part 6: Lists

## A list holds a sequence of values

````python
primes = [2, 3, 5, 7, 11]
print(len(primes))
print(primes[0], primes[-1])
print(primes[1:3])
print(7 in primes)
print(sum(primes), min(primes), max(primes))
````

Indexing, slicing, `len` and `in` work exactly as for strings.
A list can hold values of any type, even mixed: `[1, "two", 3.0]`.

## Lists can be changed

````python
primes = [2, 3, 5, 7, 11]
primes[0] = 1
primes.append(13)
print(primes)
````

Unlike strings, lists are **mutable**: methods like `append` change the list itself.

**Predict:** what's printed?

````python
a = [1, 2, 3]
b = a
b.append(4)
print(a)
````

`b = a` doesn't copy the list. Both names are labels on **the same** list,
so changing it through `b` is visible through `a`.

Remember this one. It will bite you, probably in two weeks.
Compare with `a = 1; b = a; a = 2` earlier: there we moved a label, here we changed the object.

*Try it on pythontutor.com.*

**Predict:** what's printed?

````python
numbers = [3, 1, 2]
result = numbers.sort()
print(numbers)
print(result)
````

Methods that change a list usually return `None` ("nothing").
`sorted(numbers)` instead returns a **new** sorted list and leaves `numbers` alone.

## From strings to lists and back

````python
words = "the quick brown fox".split()
print(words)
print(len(words))
print("-".join(words))
print(list("abc"))
````

## Sets: no duplicates, no order

A quick preview, useful for this week's exercise.

````python
colors = {"red", "green", "red", "blue"}
print(colors)
print(len(colors))

a = {1, 2, 3, 4}
b = {3, 4, 5}
print(a & b)
print(set([1, 1, 2, 3, 3]))
````

`&` gives the elements in both sets. `set(some_list)` removes duplicates
(and loses the order).

# Part 7: A first look at functions

## A function: a name for a piece of code

We'll look at functions properly next week. For now, just the form:

````python
def fahrenheit_to_celsius(fahrenheit):
    celsius = (fahrenheit - 32) * 5 / 9
    return celsius


print(fahrenheit_to_celsius(212))
print(fahrenheit_to_celsius(32))
````

- `def` gives the function a name and says which input it takes (`fahrenheit`)
- The **indented** lines are its body; they run when the function is **called**
- `return` hands the result back to the caller

In this week's exercise, the first line of each function is already written.
You fill in the body, and end it with `return`.

## Tests call your functions

A test calls a function with a known input and checks the result with `assert`:

````python
assert fahrenheit_to_celsius(212) == 100
assert fahrenheit_to_celsius(32) == 0
print("all good")
````

**Predict:** what happens?

````python
assert fahrenheit_to_celsius(0) == -17.8
````

A failing `assert` raises an `AssertionError`. That's what `pytest` looks for.
(And yes: never compare floats with `==`. We'll see how pytest handles that next week.)

## Summary

- Code runs **top to bottom**; an error stops it; a syntax error stops it before it starts
- Values have **types**; expressions are **evaluated** to values
- Names are **labels** on objects; assignment evaluates the right side first
- Strings are **immutable**: methods return new strings
- Lists are **mutable**: two names can label the same list
- Functions take input and `return` a result; tests call them and `assert` on the result
